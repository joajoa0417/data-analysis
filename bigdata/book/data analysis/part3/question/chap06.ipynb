##### 1. 한 커피 제조회사에서는 새로 출시한 커피의 카페인 함량이 평균 95mg 미만이라고 주장했다. 그 주장이 사실인가를 알아보기 위해 25개의 커피 샘플을 무작위로 추출했다. 커피 제조회사의 주장이 타당한지를 유의수준 5%에서 검정하시오.

- H0: μ ≥ 95mg
- H1: μ < 95mg

```python
import pandas as pd

df = pd.DataFrame({
    'Caffeine(mg)': [
        94.2, 93.7, 95.5, 93.9, 94.0, 95.2, 94.7, 93.5, 92.8, 94.4,
        93.8, 94.6, 93.3, 95.1, 94.3, 94.9, 93.9, 94.8, 95.0, 94.2,
        93.7, 94.4, 95.1, 94.0, 93.6
    ]
})
```

1-1. 표본 데이터의 평균을 구하시오.

```python
print(df['Caffeine(mg)'].mean())
# print(df.mean())
```

1-2. Shapiro-Wilk 검정의 p-value를 구하시오.

```python
from scipy import stats

stats.shapiro(df['Caffeine(mg)'])
```

1-3. 단일 표본 t-검정의 검정 통계량을 구하시오.


1-4. 단일 표본 t-검정의 p-value 를 구하시오.


1-5. 유의수준 0.05 하에서 귀무가설을 기준으로 검정 결과를 채택/기각 중 선택해 입력하시오.

```python
static, pvalue = stats.ttest_1samp(df['Caffeine(mg)'], 95, alternative='less') # 대립가설 기준
print(static)
print("{:.10f}".format(pvalue))

# 검정 통계량 : statistic=-5.501737036221897
# p-value : pvalue=5.8686553916715e-06
# p-value < 0.05 이므로 귀무가설을 기각한다.
```

##### 2. 스마트폰 배터리의 충전 시간을 줄이기 위한 새로운 충전기를 개발했다. 개발된 충전기의 효과를 검정하기 위해 스마트폰 사용자 집단을 두 그룹으로 나눠 한 그룹에는 새로운 충전기를 제공하고, 다른 그룹에는 기존의 충전기를 제공했다. 그 후 두 그룹의 평균 충전 완료 시간을 비교했다. 두 그룹 모두 충전 시간은 정규 분포를 따르며 분산은 같다고 가정한다. 새로운 충전기가 더 빠르다고 할 수 있는지를 유의수준 5%하에서 검정하시오. 여기서 두 집단은 모두 정규 분포를 따르고 분산은 같다고 가정한다. (μ₁: 새로운 충전기 집단의 모평균, μ₂: 기존 충전기 집단의 모평균)


- H0: μ₁ ≥ μ₂
- H1: μ₁ < μ₂

2-1. 위의 가설을 검정하기 위한 검정 통계량을 구하시오.


2-2. 위의 통계량에 대한 p-value를 구하시오.

2-3. 유의수준 0.05하에서 귀무가설을 기준으로 검정의 결과를 채택/기각 중 선택해 입력하시오.


```python
## 데이터
import pandas as pd

df = pd.DataFrame({
    '충전기': ['New'] * 10 + ['Old'] * 10,
    '충전시간': [
        1.5, 1.6, 1.4, 1.7, 1.5, 1.6, 1.7, 1.4, 1.6, 1.5,
        1.7, 1.8, 1.7, 1.9, 1.8, 1.7, 1.8, 1.9, 1.7, 1.6
    ]
})
print(df.head(2))
```

```python
## 독립 표본 검정
df_old = df[df['충전기'] == 'Old']
df_new = df[df['충전기'] == 'New']

from scipy import stats

stats.ttest_ind(df_new['충전시간'], df_old['충전시간'], alternative='less', equal_var=True)

# 검정통계량 statistic : -4.582575694955849
# p-value : 0.00011546547787696304
# p-value < 0.05 이므로 귀무가설 기각
```

- μ = (새로운 교수 방법 - 기존 교수 방법)의 평균
- H0: μ = 0
- H1: μ < 0

```python
## 데이터
import pandas as pd

df = pd.DataFrame({
    'User': list(range(1, 11)),
    '기존방법': [60.4, 60.7, 60.5, 60.3, 60.8, 60.6, 60.2, 60.5, 60.7, 60.4],
    '새로운방법': [59.8, 60.2, 60.1, 59.9, 59.7, 58.4, 57.0, 60.3, 59.6, 59.8]
})
print(df.head(2))
```

3-1. μ의 표본 평균을 구하시오.


3-2. 위의 가설을 검정하기 위한 검정 통계량을 구하시오.


3-3. 위의 통계량에 대한 p-value를 구하시오.


3-4. 유의수준 0.05에서 귀무가설을 기준으로 검정의 결과를 채택/기각 중 선택해 입력하시오.

```python
## 대응표본 검정
# 3-1.
df['diff'] = df['새로운방법'] - df['기존방법']
print(df['diff'].mean())

from scipy import stats

# 3-2 ~ 3-4.
print(stats.ttest_rel(df['새로운방법'], df['기존방법'], alternative='less'))
# statistic = -3.407973078114844
# p-value = 0.0038872633380070652
# p-value < 0.05이므로 귀무가설을 기각한다.
# 새로운 교육 방법이 교육 시간을 단축시킨다.
```

##### 4. 4개의 중학교에서 다른 교육 방법을 사용해 학생들에게 수학을 가르쳤다. 각 중학교의 학생들이 받은 교육 후의 수학 성적은 다음과 같다. 이 4개의 교육 방법에 따른 성적 차이가 통계적으로 유의미한지 5%의 유의수준에서 검정하시오.

- H0 : 모든 중학교의 학생들의 수학 성적 평균은 동일하다.
- H1 : 적어도 2개 중학교의 학생들의 수학 성적 평균은 다르다.

```python
import pandas as pd

# df = pd.read_csv("math.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part3/ch6/math.csv")
print(df.head())
```

4-1. 각 그룹의 수학 성적에 대해 Shapiro-Wilk 검정을 통한 정규성을 확인하고, 그 결과의 p-value를 구하시오.

4-2. 4개 그룹의 수학 성적이 등분산성을 갖는지 확인하기 위해 Levene 검정을 실시하고, p-value를 구하시오.

4-3. 유의수준 0.05 하에서 귀무가설을 기준으로 검정의 결과를 채택/기각 중 선택해 입력하시오.

4-4. 그룹 변수의 자유도를 구하시오.

4-5. 잔차의 자유도를 구하시오.

4-6. 성적의 제곱합을 구하시오.

4-7. 성적의 평균 제곱을 구하시오.

4-8. F-통계량의 값을 구하시오.

4-9. 성적에 대한 p-value를 구하시오.

```python
from scipy import stats

# 4-1.
print('====== 정규성 검정 ======')
a = df[df['groups'] == 'group_A']
b = df[df['groups'] == 'group_B']
c = df[df['groups'] == 'group_C']
d = df[df['groups'] == 'group_D']

print(stats.shapiro(a['scores'])) # 정규성
print(stats.shapiro(b['scores'])) # 정규성
print(stats.shapiro(c['scores'])) # 정규성
print(stats.shapiro(d['scores'])) # 정규성

# 4-2.
print('\n====== 등분산성 검정 ======')
print(stats.levene(a['scores'], b['scores'], c['scores'], d['scores'])) # 등분산성

# 4-3. 
print('\n====== 일원 분산 분석 ======')
static, pval = stats.f_oneway(a['scores'], b['scores'], c['scores'], d['scores'])
print('statistic:', static)
print("p-value:", "{:.10f}".format(pval))
# p-value < 0.05 이므로 귀무가설 기각

# 4-4 ~ 4-9.
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm

print('\n====== ANOVA 테이블 출력 ======')
model = ols('scores ~ groups', data=df).fit()
print(anova_lm(model))
# group 자유도 : 3
# 잔차의 자유도 : 36
# 성적의 제곱합 : 411.8
# 성적의 평균 제곱 : 137.266667
# F-통계량 : 34.174274
# 성적의 p-value : 1.240642e-10
```
