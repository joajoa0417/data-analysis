##### 작업형 제 1유형

```python
import pandas as pd
df = pd.read_csv('student_assessment.csv')

df = df.dropna()
id_max = df['id_assessment'].value_counts(ascending=False).idxmax()

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df1 = df[df['id_assessment'] == id_max]
df1['score'] = scaler.fit_transform(df1[['score']])
round(df1['score'].sort_values(ascending=False).max(), 3)
```

##### 작업형 제 3유형

```python
# 1
import pandas as pd
df = pd.read_csv('clam.csv')

train = df.iloc[:210]
test = df.iloc[210:]
# train.shape, test.shape

import numpy as np
from statsmodels.formula.api import logit
model = logit('gender ~ weight', data=train).fit()
print('odds ratio:', round(np.exp(model.params['weight']), 4))

model2 = logit('gender ~ age + length + diameter + height + weight', data=train).fit()
print('잔차이탈도:', round(-2 * model2.llf, 2))

from sklearn.metrics import accuracy_score
target = test.pop('gender')
pred = model.predict(test)
pred = (pred > 0.5).astype(int)
acc = accuracy_score(target, pred)
# print('accuracy-score:', acc)
print('error-ratio:', round(1 - acc, 3))
```

```python
import pandas as pd
df = pd.read_csv('system_cpu.csv')

print(round(df.corr()['ERP'].iloc[1:].sort_values(ascending=False).max(), 3))

from statsmodels.formula.api import ols
df1 = df[df['CPU'] < 100]
model = ols('ERP ~ Feature1 + Feature2 + Feature3 + CPU', data=df1).fit()
# print(model.summary())
print(round(model.rsquared, 3))
print(round(model.pvalues.max(), 3))
```
