##### 작업형 제 1유형

```python
# 1
import pandas as pd
df = pd.read_csv('drinks.csv')

df_pivot = df.pivot_table(index='continent', values='beer_servings', aggfunc='mean')
df_pivot = df_pivot.sort_values('beer_servings', ascending=False)
max_cont = df_pivot['beer_servings'].idxmax()
df_cont = df[df['continent'] == max_cont]
df_cont.sort_values('beer_servings', ascending=False).iloc[4]['beer_servings']
```

```python
# 2 
import pandas as pd
df = pd.read_csv('tourist.csv')
df['sum'] = df.sum(axis=1, numeric_only=True)
df['관광객비율'] = df['관광'] / df['sum']
df1 = df.sort_values('관광객비율', ascending=False)
a = df1.iloc[1]['사업']
b = df.sort_values('관광', ascending=False).iloc[1]['공무']
print(a + b)
```

##### 작업형 제 3유형

```python
# 1
import pandas as pd
df = pd.read_csv('churn.csv')

from statsmodels.formula.api import logit
model = logit('Churn ~ AccountWeeks	+ ContractRenewal \
                + DataPlan + DataUsage + CustServCalls + DayMins \
                + DayCalls + MonthlyCharge + OverageFee + RoamMins', data=df).fit()
# print(model.summary())

print('유의하지 않은 독립변수 개수:', sum(model.pvalues[1:] > 0.05))
model = logit('Churn ~ DataUsage + DayMins', data=df).fit()
# print(model.summary())
print('유의한 회귀계수 합계:', round(sum(model.params), 3))

import numpy as np
print('오즈비:', round(np.exp(model.params['DataUsage'] * 5), 3))
```

```python
# 2
import pandas as pd
df = pd.read_csv('piq.csv')

from statsmodels.formula.api import ols
model = ols('PIQ ~ Brain + Height + Weight', data=df).fit()
# print(model.summary())
p_min = (model.pvalues).sort_values(ascending=True).idxmin()
print('유의확률이 가장 작은 변수의 회귀 계수 값:', round(model.params[p_min], 3))
print('결정계수(R-squared):', round(model.rsquared, 2))

new_data = pd.DataFrame({'Brain':[90], 'Height':[70], 'Weight':[150]})
pred = model.predict(new_data)
print('예측 PIQ:', round(pred[0]))
```
